```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)

n_users = 30000

group = np.random.choice(["control", "treatment"], size=n_users, p=[0.5, 0.5])

conversion_prob = np.where(group == "control", 0.22, 0.25)
converted = np.random.binomial(1, conversion_prob)

AOV_mean = np.where(group == "control", 1000, 1400)
AOV_std = np.where(group == "control", 5000, 5800)
AOV_col = np.random.normal(loc=AOV_mean, scale=AOV_std)

platform = np.random.choice(["android", "ios", "web"], size=n_users, p=[0.25, 0.45, 0.3])

df = pd.DataFrame({
    "user_id": range(1, n_users + 1),
    "group": group,
    "converted": converted,
    "AOV_col": AOV_col,
    "platform": platform
})
df.to_csv("../data/ab_test_dataset.csv", index=False)
print(df.head())
print(df.groupby("group")["converted"].agg(["count", "mean"]))
```

# Пояснения к блокноту

Генерируем датасет для дальнейшего тестирования с основной метрикой конверсии "converted" и guardrail метрикой "AOV_col", где в изначальной A группе конверсия 0.22, AOV с средним 1000 и стандартным отклонением 10, в B группе конверсия 0.25, AOV с средним 1400 и стандартным октлонением 12.

"platform" добавлено для дополнительной сегментации.
