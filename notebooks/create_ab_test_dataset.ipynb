Далее "control" группа со старой лентой без ключевых характеристик, "treatment" с новой лентой с ключевыми характеристиками.

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)

n_users = 30000

group = np.random.choice(["control", "treatment"], size=n_users, p=[0.5, 0.5])

conversion_prob = np.where(group == "control", 0.22, 0.25)
converted = np.random.binomial(1, conversion_prob)

ACV_mean = np.where(group == "control", 1000, 1015)
ACV_std = np.where(group == "control", 4000, 4200)
ACV_col = np.random.normal(loc=ACV_mean, scale=ACV_std)

platform = np.random.choice(["android", "ios", "web"], size=n_users, p=[0.25, 0.45, 0.3])

df = pd.DataFrame({
    "user_id": range(1, n_users + 1),
    "group": group,
    "converted": converted,
    "ACV_col": ACV_col,
    "platform": platform
})
df.to_csv("../data/ab_test_dataset.csv", index=False)
print(df.head())
print(df.groupby("group")["converted"].agg(["count", "mean"]))
```

# Пояснения к блокноту

Генерируем датасет для дальнейшего тестирования с основной метрикой конверсии "converted" и guardrail метрикой "ACV_col" (ACV — цена товара, добавленного в корзину), где в изначальной A группе конверсия 0.22, ACV со средним 1000 и стандартным отклонением 10, в B группе конверсия 0.25, ACV со средним 1400 и стандартным отклонением 12.

"platform" добавлено для дополнительной сегментации.
