```python
import pandas as pd
from scipy.stats import chisquare
```

```python
df = pd.read_csv("../data/ab_test_dataset.csv")

observed = df["group"].value_counts()
print(observed)

expected_len = len(df) / 2
chi2, p_value_srm = chisquare(observed, [expected_len, expected_len])
print(f"Chi2: {chi2:.3f}, p-value: {p_value_srm:.4f}")

if p_value_srm < 0.01:
    print("SRM нет.")
else:
    print("SRM есть.")
```

# Пояснения про SRM

Из за ошибок при разбиении на группы может появится SRM (диспропорция относительно изначально заданного значения), если значение $p\text{-}value < 0.01$ при тесте хи-квадрат, то SRM присутствует. 

При обноружении SRM следует исправить деление на группы, так как дальнейшие тесты работают при определённом значении пропорций между A и B группой.
